```python
import torch
import torch.nn as nn
```

```python
#input
torch.manual_seed(42)
inputs = torch.tensor([1.0,2.0,3.0])
print(inputs)
```

```python
#activation fn
#we dont have the support of stepn fn so we are going to use something we call as linear activation fn
#Linear activation fn we pass the no. of ip_feat and op_feat
neuron = nn.Linear(in_features=3,out_features=1)
```

```python
# now to get the output
output = neuron(inputs)
print(output)
```

```python
#verify formula
neuron.weight @ inputs + neuron.bias
```

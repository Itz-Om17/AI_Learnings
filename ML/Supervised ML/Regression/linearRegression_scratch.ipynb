```python
import numpy as np
```

```python
class LinearRegression:
  def __init__(self,learning_rate=0.01,n_itr = 1000):
    self.bias = None
    self.weights = None
    self.lr = learning_rate
    self.n_itr = n_itr

  def fit(self,X,y):
    m,n = X.shape #this will willk give no. of rows and features value
    #step1
    self.bias = 0
    self.weights = np.zeros(n) #zeros should be for no. of features we may have X1θ1+...+Xnθn so the bais zeros should be also that much

    for i in range (self.n_itr):
      #step2
      y_pred = self.bias + np.dot(X, self.weights)

      #step3
      #derivaitve w.r.t. bias(θ0)
      db = (1/m) * np.sum(y_pred-y)
      #derivaitve w.r.t. weights(θ1,...θn)
      dw = (1/m)*np.dot(X.T, (y_pred-y))

      #step4
      self.bias = self.bias - self.lr * db
      self.weights = self.weights - self.lr * dw



  def predict(self, X):
    y_pred = self.bias + np.dot(X, self.weights)
    return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()

model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weights)
```
